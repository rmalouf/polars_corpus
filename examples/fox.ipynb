# Keyword analysis

```python
import polars as pl
import polars_corpus as plc
from scipy.stats import spearmanr

import matplotlib.pyplot as plt
import seaborn as sns

pl.Config.set_tbl_rows(20)
pl.Config.set_float_precision(3)
pl.Config.set_thousands_separator(",")
```

```python
pl.Config.set_tbl_rows(25)
pl.Config.set_float_precision(3)
```

```python
coca = pl.scan_parquet("/Volumes/Corpora/COCA/coca.parquet") \
    .filter(pl.col('genre')=='SPOK')
```

```python
target_type = "SPOK:FOX"

tt_fox_keywords = plc.keywords(
    target=coca.filter(pl.col("text_type") == target_type),
    reference=coca.filter(pl.col("text_type") != target_type),
    method="ttest",
    expr=pl.col("token").str.to_lowercase(),
).collect(engine="streaming")
```
