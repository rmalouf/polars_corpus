```python
import polars as pl
import polars.selectors as cs
import polars_corpus as plc
```

```python
c = pl.read_parquet('bnc.parquet') #.head(5000000)
```

```python
r = plc.search(c, '[c5="AJ.*"]+ [c5="NN.*"]')
r
```

```python
r = plc.search(c, '[pos="SUBST"]{3,}')
```

```python
r.matches() \
    .with_columns(pl.col('token').list.join(' ')) \
    .group_by('token') \
    .len() \
    .sort(by='len', descending=True)
```

```python
%%time

left, kwic, right = [ ], [ ], [ ]
for s, e in r.matched_spans:
    s2 = max(s-5, 0)
    e2 = e + 5
    context = list(c['token'].slice(s2, e2-s2))
    left.append(context[:s-s2])
    kwic.append(context[s-s2:-5])
    right.append(context[-5:])
    #left.append(c['token'].slice(s-5, 5).implode()[0])
    #kwic.append(c['token'].slice(s, e-s).implode()[0])
    #right.append(c['token'].slice(e, 5).implode()[0])
concordance = pl.DataFrame({'left': left, 'kwic': kwic, 'right': right})
```
