```python
import polars as pl
import polars_corpus as plc
```

```python
c = pl.read_parquet('bnc.parquet') #.head(5000000)
```

```python
r = plc.search(c, '[c5="AJ.*"]+ [c5="NN.*"]')
r
```

```python
r.matches().group_by('token').len().sort(by='len', descending=True)
```

```python
r = plc.search(c, '[pos="SUBST"]{3,}')
```

```python
r.matches() \
    .with_columns(pl.col('token').list.join(' ')) \
    .group_by('token') \
    .len() \
    .sort(by='len', descending=True)
```
